```python
# At the top of this decision tree is the root node or the parent node. The root node represents the entire population. This node gets divided into two or more homogeneous sets in our decision tree to answer the questions which is a branch or subsection of the tree. When we divide a node into two or more subnodes, it's called splitting.

# When we split the root node into two subnodes, they are called child nodes which split again into subnodes if another question in them (decision node). The four results do not split further and are referred to as a leaf or terminal node.
```

```python
# Initial imports
import pandas as pd
from path import Path
from sklearn import tree
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
# Our goal is to predict if a loan application is worthy of approval based on information we have in our df_loans DataFrame.
```

## Loading and Preprocessing Loans Encoded Data

```python
# Loading data
file_path = Path("../Resources/loans_data_encoded.csv")
df_loans = pd.read_csv(file_path)
df_loans.head()
```

```python
# Define features set
X = df_loans.copy()
X = X.drop("bad", axis=1)
X.head()
```

```python
#The target set is the bad column, indicating whether or not a loan application is good (0) or bad (1). 
```

```python
# Define target vector
y = df_loans["bad"].values.reshape(-1, 1)
y[:5]
```

```python
# Splitting into Train and Test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=78)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
#We can manually specify the desired split with the train_size parameter.
```

```python
print(X_train2.shape)
print(X_test2.shape)
print(y_train2.shape)
print(y_test2.shape)
```

```python
#By default, our training and testing data sets are 75% and 25%, respectively, of the original data. Using the following code, we can see the data's 75-25 split.

#The output from running the code above shows that the X_train and y_train is 75% of 500 and that the X_test and y_test are 25%.
```

```python
# Creating StandardScaler instance
scaler = StandardScaler()
```

```python
# Fitting Standard Scaller
X_scaler = scaler.fit(X_train)
```

```python
# Scaling data
X_train_scaled = X_scaler.transform(X_train)
X_test_scaled = X_scaler.transform(X_test)
```

## Fitting the Decision Tree Model

```python
# Creating the decision tree classifier instance
model = tree.DecisionTreeClassifier()
```

```python
# Fitting the model
model = model.fit(X_train_scaled, y_train)
```

## Making Predictions Using the Tree Model

```python
# Making predictions using the testing data
predictions = model.predict(X_test_scaled)
```

```python
#The output from this code will be an array of 125 predictions with either a 1 for a bad loan application or a 0 for a good, or approved, loan application.
```

```python
predictions
```

## Model Evaluation

```python
# Calculating the confusion matrix
cm = confusion_matrix(y_test, predictions)
cm_df = pd.DataFrame(
    cm, index=["Actual 0", "Actual 1"], columns=["Predicted 0", "Predicted 1"]
)

cm_df
```

```python
# The CM results show that:

# Out of 84 good loan applications (Actual 0), 52 were predicted to be good (Predicted 0), which we call true positives.
# Out of 84 good loan applications (Actual 0), 32 were predicted to be bad (Predicted 1), which are considered false negatives.
# Out of 41 bad loan applications (Actual 1), 21 were predicted to be good (Predicted 0) and are considered false positives.
# Out of 41 bad loan applications (Actual 1), 20 were predicted to be bad (Predicted 1) and are considered true negatives.
```

```python
# 0 - good
# 1 - bad
```

```python
# Calculating the accuracy score
acc_score = accuracy_score(y_test, predictions)
```

```python
acc_score
```

```python
#(True Positives (TP) + True Negatives (TN)) / Total = (52 + 20)/125
x = cm_df["Predicted 0"][0]+cm_df["Predicted 1"][1]
total = cm_df["Predicted 0"][0]+cm_df["Predicted 1"][1]+cm_df["Predicted 0"][1]+cm_df["Predicted 1"][0]
y = x/total
y
```

```python
# Displaying results
print("Confusion Matrix")
display(cm_df)
print(f"Accuracy Score : {acc_score}")
print("Classification Report")
print(classification_report(y_test, predictions))
```

```python
# Let's go over the results in the classification report:

# Precision: Precision is the measure of how reliable a positive classification is. From our results, the precision for the good loan applications can be determined by the ratio TP/(TP + FP), which is 52/(52 + 21) = 0.71. The precision for the bad loan applications can be determined as follows: 20/(20 + 33) = 0.38. A low precision is indicative of a large number of false positives. Of the 52 loan applications we predicted to be bad applications, 32 were actually good loan applications.
# Recall: Recall is the ability of the classifier to find all the positive samples. It can be determined by the ratio: TP/(TP + FN), or 52/(52 + 32) = 0.62 for the good loans and 20/(21 + 20) = 0.49 for the bad loans. A low recall is indicative of a large number of false negatives.
# F1 score: F1 score is a weighted average of the true positive rate (recall) and precision, where the best score is 1.0 and the worst is 0.0.
# Support: Support is the number of actual occurrences of the class in the specified dataset. For our results, there are 84 actual occurrences for the good loans and 41 actual occurrences for bad loans.
# In summary, this model may not be the best one for preventing fraudulent loan applications because the model's accuracy, 0.576, is low, and the precision and recall are not good enough to state that the model will be good at classifying fraudulent loan applications. Modeling is an iterative process: you may need more data, more cleaning, another model parameter, or a different model. It's also important to have a goal that's been agreed upon, so that you know when the model is good enough.
```
